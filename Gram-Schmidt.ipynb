The Gram-Schmidt process is used to create an orthonormal basis from a set of linearly independent vectors. This process onsists of two steps: orthogonalization, and normalization. Orthogonalization is done by subtracting the component of a vector in the direction of the previous vectors, which results in a vector that is orthogonal to all previous vectors. This step alone gives us an orthogonal basis. Normalization is what makes it an orthonormal basis, and it is simply done by dividing each vector by its norm. 

```python
import numpy as np
import numpy.linalg as la

def gsprocess(A) : #A is a matrix
    B = np.array(A, dtype=np.float64) # Making B as a copy of A, since we're going to alter its values 
    # looping through the columns in the matrix, i.e. the vectors
    for i in range(B.shape[1]) :
        # looping through all previous vectors up until the ith vector
        for j in range(i) :
            # Subtracting the overlap with previous vectors (orthogonalization) 
            B[:, i] = B[:, i] - B[:, i] @ B[:, j] * B[:, j]
        # normalizing the ith vector 
        if la.norm(B[:, i]) > 1e-14: # making sure the ith vector is linearly independent of other vectors
            B[:, i] = B[:, i]/la.norm(B[:, i])
        else: # otherwise setting the vector to zero. it cannot be a vector in the orthonormal basis 
            B[:, i] = np.zeros_like(B[:, i])
    return B 
```
